# Exploración de datos sintéticos | ReviewPulse 2.0\nMuestra ciega de 120 reseñas. Las etiquetas manuales solo se analizan cuando exista `gold.csv`. No abrir la verdad del simulador.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

RAIZ = Path.cwd()
if not (RAIZ / "data/gold/muestra.csv").exists():
    RAIZ = RAIZ.parent
muestra = pd.read_csv(RAIZ / "data/gold/muestra.csv")
assert list(muestra.columns) == ["id", "texto"] and len(muestra) == 120
muestra["longitud_caracteres"] = muestra["texto"].str.len()
muestra["longitud_palabras"] = muestra["texto"].str.split().str.len()
display(muestra[["id", "longitud_caracteres", "longitud_palabras"]].describe(include="all"))
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(muestra["longitud_palabras"], bins=12)
ax.set(title="Distribución de longitud de reseñas", xlabel="Palabras", ylabel="Frecuencia")
plt.show()
```

```python
gold = RAIZ / "data/gold/gold.csv"
if gold.exists():
    etiquetas = pd.read_csv(gold)
    assert len(etiquetas) == 120 and set(etiquetas.columns) == {"id", "texto", "polaridad", "tema"}
    fig, ax = plt.subplots(1, 2, figsize=(11, 4))
    etiquetas["tema"].value_counts().sort_index().plot(kind="bar", ax=ax[0], title="Temas etiquetados")
    etiquetas["polaridad"].value_counts().sort_index().plot(kind="bar", ax=ax[1], title="Polaridades etiquetadas")
    plt.tight_layout(); plt.show()
else:
    print("Distribuciones por tema y polaridad: pendientes hasta consolidar el etiquetado humano.")

print("Distribución por plataforma: no disponible (el generador no registra esa variable).")
```
